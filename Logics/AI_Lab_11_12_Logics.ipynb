```python
from logics import vars
```

```python
P, Q, A, B, C = vars('P', 'Q', 'A', 'B', 'C')
```

```python
# P v ¬P
exp = (P | ~P)
exp.print_truth_table()
exp.printValidity()
exp.printSatisfiability()
```

```python
# P → P
exp = (P >> P)
exp.print_truth_table()
exp.printValidity()
exp.printSatisfiability()
```

```python
# (A ∧ B) ↔ (B → C)
exp = ((A & B).iff(B >> C))
exp.print_truth_table()
exp.printValidity()
exp.printSatisfiability()
```

```python
# P → Q ⊨ ¬ P → ¬ Q
kb = (P >> Q)
alpha = (~P >> ~Q)
r = (kb >> alpha)

r.print_truth_table()
r.printSatisfiability()
if r.isPropValid():
    print(f"{kb} entails {alpha}")
else:
    print(f"{kb} does not entail {alpha}")
```
