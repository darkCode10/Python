```python
import random
```

```python
##Step 1
chromosomes=["101011","001011","110001","011011"]
decimal_values=[]
for x in chromosomes:
    y=int(x,2)
    decimal_values.append(y)
print(decimal_values)
```

```python
##Step 2
fitness_values=[]
for x in decimal_values:
    raw_fitness=(x/63)*2-1
    fitness_per=(raw_fitness+1)*50
    fitness_values.append(fitness_per)
print(fitness_values)
```

```python
##Step 3
best_fitness=max(fitness_values)
best_index=fitness_values.index(best_fitness)
best_chromosome=chromosomes[best_index]
print(f"Best Chromosomes: {best_chromosome} with Fitness: {best_fitness}")
```

```python
##Step 4 roulette wheel selection
selection_prob=[]
for x in fitness_values:
    y=x/sum(fitness_values)
    selection_prob.append(y)
print(selection_prob)
```

```python
parent1,parent2=random.choices(chromosomes,weights=selection_prob,k=2)
print(parent1,parent2)
```

```python
##Step 5 crossover (single point)
crossover_point=random.randint(1,len(parent1)-1)
print(crossover_point)
child1=parent1[:crossover_point]+parent2[crossover_point:]
child2=parent2[:crossover_point]+parent1[crossover_point:]
print(child1,child2)
```

```python
## (Crossover multipoint)
point1,point2=sorted(random.sample(range(1,len(parent1)-1),2))
print(point1,point2)
print(parent1,parent2)
child1=parent1[:point1]+parent2[point1:point2]+parent1[point2:]
child2=parent2[:point1]+parent1[point1:point2]+parent2[point2:]
print(child1,child2)
```

```python
##Step 6 Mutation
mutation_rate=0.2
def mutate(chromosome):
    chromosome_list=list(chromosome)
    for x in range(len(chromosome_list)):
        if random.random()<mutation_rate:
            if chromosome_list[x]=='0':
                chromosome_list[x]='1'
            else:
                chromosome_list[x]='0'
    return ''.join(chromosome_list)
child1=mutate(child1)
child2=mutate(child2)
print(child1,child2)
```

```python
##Step 7 Create Generation
chromosomes_with_fitness=list(zip(chromosomes,fitness_values))
chromosomes_with_fitness.sort(key=lambda x:x[1],reverse=True)
sorted_chromosomes=[]
for x,y in chromosomes_with_fitness:
    sorted_chromosomes.append(x)
    chromosomes=sorted_chromosomes[:-2]+[child1,child2]
print(sorted_chromosomes)
print(chromosomes)
```
